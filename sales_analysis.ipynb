```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Read CSV file
df = pd.read_csv('Superstore.csv', encoding='latin1')
df.drop_duplicates(inplace=True)

# 2. Chart 1: Total Sales vs Profit by Category
cat_summary = df.groupby('Category')[['Sales', 'Profit']].sum().reset_index()
cat_melted = cat_summary.melt(id_vars='Category', var_name='Metric', value_name='Amount')

plt.figure(figsize=(8, 5))
sns.barplot(data=cat_melted, x='Category', y='Amount', hue='Metric', palette='viridis')
plt.title('Total Sales vs. Profit by Category')
plt.ylabel('Amount (USD)')
plt.tight_layout()
plt.savefig('category_sales_profit.png', dpi=300)
plt.show()

print("Cell 1 executed: Category chart saved!")
```

```python
# Chart 2: Net Profit by Sub-Category
sub_summary = df.groupby('Sub-Category')['Profit'].sum().reset_index()
sub_summary = sub_summary.sort_values(by='Profit', ascending=True)

plt.figure(figsize=(10, 6))
colors = ['red' if x < 0 else 'green' for x in sub_summary['Profit']]
plt.barh(sub_summary['Sub-Category'], sub_summary['Profit'], color=colors)
plt.title('Net Profit by Sub-Category')
plt.xlabel('Profit (USD)')
plt.ylabel('Sub-Category')
plt.axvline(0, color='black', linewidth=0.8, linestyle='--')
plt.tight_layout()
plt.savefig('subcategory_profit.png', dpi=300)
plt.show()

print("Cell 2 executed: Sub-Category profit chart saved!")
```

```python
# Chart 3: Impact of Discount on Profitability
plt.figure(figsize=(9, 5))
sns.scatterplot(data=df, x='Discount', y='Profit', hue='Category', palette='deep', alpha=0.7)
plt.title('Impact of Discount on Profitability')
plt.xlabel('Discount Rate')
plt.ylabel('Profit (USD)')
plt.axhline(0, color='red', linestyle='--', linewidth=0.8)
plt.tight_layout()
plt.savefig('discount_profit.png', dpi=300)
plt.show()

print("Cell 3 executed: Discount impact chart saved!")
```

```python
# Chart 4: Total Sales by Ship Mode
ship_summary = df.groupby('Ship Mode')['Sales'].sum().reset_index()

plt.figure(figsize=(7, 4))
sns.barplot(data=ship_summary, x='Ship Mode', y='Sales', palette='Blues_d')
plt.title('Total Sales by Ship Mode')
plt.xlabel('Ship Mode')
plt.ylabel('Total Sales (USD)')
plt.tight_layout()
plt.savefig('shipmode_sales.png', dpi=300)
plt.show()

print("Cell 4 executed: Ship Mode sales chart saved successfully!")
```
